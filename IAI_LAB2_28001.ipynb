```python
import numpy as np
from scipy.stats import stats
import math

n= int(input("Enter the value of N : "))
print(n)

p = float(input("Enter the value of P : "))
print(p)

k = int(input("Specific number of successes you're interested in : "))
print(k)

def binomial_probability(n, p, k):
  combinations = math.comb(n, k)
  probability = combinations * (p ** k) * ((1 - p) ** (n - k))
  return probability

probabilities = []
for k in range(n+1):
  probability = binomial_probability(n, p, k)
  probabilities.append(probability)
  print(print(f"Heads: {k} | Probability: {prob:.6f}"))

expected_heads = n*p
expected_tails = n*(1-p)

print("Expected Heads : ", expected_heads)
print("Expected Tails : ", expected_tails)
```
