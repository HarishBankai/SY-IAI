```python
import random

def heuristic(state, goal):
    score = 0
    goal_support = {}
    for stack in goal:
        for i, block in enumerate(stack):
            goal_support[block] = stack[i-1] if i > 0 else "TABLE"

    for stack in state:
        for i, block in enumerate(stack):
            current_support = stack[i-1] if i > 0 else "TABLE"
            if block in goal_support:
                if current_support == goal_support[block]:
                    score += 1
                else:
                    score -= 1
    return score

def generate_neighbors(state):
    neighbors = []
    for i in range(len(state)):
        if not state[i]: continue
        for j in range(len(state)):
            if i == j: continue
            new_state = [list(s) for s in state]
            block = new_state[i].pop()
            new_state[j].append(block)
            new_state = [s for s in new_state if s]
            neighbors.append(new_state)
        if len(state[i]) > 1:
            new_state = [list(s) for s in state]
            block = new_state[i].pop()
            new_state.append([block])
            neighbors.append(new_state)
    return neighbors

def hill_climbing(initial, goal, max_steps=100):
    current = initial
    current_score = heuristic(current, goal)
    for _ in range(max_steps):
        if current == goal: return current
        neighbors = generate_neighbors(current)
        random.shuffle(neighbors)
        best_neighbor = current
        best_score = current_score
        for neighbor in neighbors:
            score = heuristic(neighbor, goal)
            if score > best_score:
                best_score = score
                best_neighbor = neighbor
            elif score == best_score and best_neighbor == current:
                best_neighbor = neighbor
        if best_neighbor == current: break
        current = best_neighbor
        current_score = best_score
    return current

def random_restart_hill_climbing(initial, goal, restarts=20):
    for _ in range(restarts):
        res = hill_climbing(initial, goal)
        if res == goal: return res
        # If stuck, scramble the state slightly and try again
        for _ in range(3):
            n = generate_neighbors(initial)
            if n: initial = random.choice(n)
    return initial

initial_state = [['C', 'A', 'D'], ['B']]
goal_state = [['B', 'C', 'A', 'D']]

solution = random_restart_hill_climbing(initial_state, goal_state)

print("Initial State:", [['C', 'A', 'D'], ['B']])
print("Final State:", solution)
print("Goal State:", goal_state)
```
